# **One Hot Encoding**

```python
import pandas as pd

B = ['bird', 'cat', 'dog', 'bird', 'bird']
d = {'categorical': B}

df = pd.DataFrame(d)
```

```python
df
```

```python
# "dummies" is used to create columns corresponding to unique values
dfDummies = pd.get_dummies(df['categorical'], prefix='category')
```
