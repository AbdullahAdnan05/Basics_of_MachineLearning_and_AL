### **Assignment: Types of Statistics and Their Applications**  

#### **Introduction**  
Statistics is a fundamental field of study that deals with the collection, analysis, interpretation, and presentation of data. It helps in decision-making and problem-solving across various domains, including business, healthcare, finance, and technology. Broadly, statistics is categorized into four main types: **Descriptive Statistics, Inferential Statistics, Predictive Statistics, and Prescriptive Statistics**. Each type serves a unique purpose and is applied in different scenarios.

---

## **1. Descriptive Statistics**  
Descriptive statistics refers to techniques that summarize and organize data to make it easier to understand. It does not involve drawing conclusions beyond the given data but provides a clear picture of the dataset.  

### **Key Components of Descriptive Statistics**  
- **Measures of Central Tendency:** Mean, Median, Mode (represent the center of the data).  
- **Measures of Dispersion:** Range, Variance, Standard Deviation (show how data is spread).  
- **Graphical Representations:** Histograms, Pie Charts, Box Plots (visualize the data distribution).  

### **Example of Descriptive Statistics**  
A teacher records the marks of students in a mathematics exam and calculates:  
- **Mean (Average Score):** 75  
- **Median (Middle Score):** 78  
- **Mode (Most Common Score):** 80  
- **Standard Deviation (Variability):** 10  

These values help the teacher understand the overall performance of the class.  

---

## **2. Inferential Statistics**  
Inferential statistics allows us to make predictions or generalizations about a larger population based on a sample. It uses probability theory and hypothesis testing to infer conclusions beyond the collected data.  

### **Key Techniques of Inferential Statistics**  
- **Hypothesis Testing:** Determines if a claim about data is statistically significant.  
- **Confidence Intervals:** Estimates a range within which a population parameter likely falls.  
- **Regression Analysis:** Examines relationships between variables.  

### **Example of Inferential Statistics**  
A pharmaceutical company tests a new drug on 500 patients (sample) and finds that it effectively reduces blood pressure. Using inferential statistics, the company estimates that the drug will be effective for the entire population of heart patients.  

---

## **3. Predictive Statistics**  
Predictive statistics uses historical data and statistical algorithms to forecast future outcomes. It is widely used in fields like finance, marketing, and machine learning.  

### **Key Techniques of Predictive Statistics**  
- **Regression Models:** Linear and logistic regression predict relationships.  
- **Time Series Analysis:** Forecasts future trends based on past data.  
- **Machine Learning Algorithms:** Uses AI-based models to predict outcomes.  

### **Example of Predictive Statistics**  
An e-commerce company analyzes customer purchase history to predict which products they might buy next, enabling personalized recommendations.  

---

## **4. Prescriptive Statistics**  
Prescriptive statistics goes beyond prediction by suggesting the best course of action based on data analysis. It incorporates optimization techniques, simulations, and decision theory.  

### **Key Techniques of Prescriptive Statistics**  
- **Optimization Models:** Finds the best possible solution for decision-making.  
- **Simulation Models:** Tests different scenarios to choose the best strategy.  
- **Decision Analysis:** Helps in strategic business decisions.  

### **Example of Prescriptive Statistics**  
A ride-sharing company like Uber analyzes traffic patterns, driver availability, and customer demand to determine the best routes and pricing strategies in real-time.  

---

## **Conclusion**  
Statistics is a powerful tool for data-driven decision-making. **Descriptive statistics** helps summarize data, **inferential statistics** allows us to make generalizations, **predictive statistics** forecasts future trends, and **prescriptive statistics** suggests optimal decisions. Understanding these types is essential for fields like data analysis, business intelligence, and artificial intelligence.

---

### 🎯 1. **Joint Probability** — $P(A \cap B)$ or $P(A, B)$

> The probability that **both events A and B happen together**.

#### Formula:

If $A$ and $B$ are two events:

$$
P(A \cap B) = P(A, B)
$$

If $A$ and $B$ are **independent**:

$$
P(A, B) = P(A) \cdot P(B)
$$

---

### 🎯 2. **Marginal Probability**

> The probability of **one event regardless of others** (e.g., total probability of A happening, no matter what B does).

#### Formula:

If you know the **joint probabilities**, then:

$$
P(A) = \sum_B P(A, B) \quad \text{(for discrete variables)}
$$

$$
P(A) = \int P(A, B) \, dB \quad \text{(for continuous variables)}
$$

This is called **marginalizing out B**.

---

### 🎯 3. **Conditional Probability** — $P(A \mid B)$

> The probability of event A **given** that B has already happened.

#### Formula:

$$
P(A \mid B) = \frac{P(A, B)}{P(B)} \quad \text{(if } P(B) > 0 \text{)}
$$

This is used *everywhere* in ML — like in Naive Bayes, Bayesian networks, etc.

---

### 🔁 Relationships Summary

* **Joint**: $P(A, B) = P(A \mid B) \cdot P(B)$
* **Conditional**: $P(A \mid B) = \frac{P(A, B)}{P(B)}$
* **Marginal**: $P(A) = \sum_B P(A, B)$

---

### 📦 Example:

Let:

* $A = \text{Rain}$
* $B = \text{Sprinkler is on}$

Say:

* $P(\text{Rain}) = 0.2$
* $P(\text{Sprinkler} \mid \text{Rain}) = 0.1$

Then:

* $P(\text{Rain, Sprinkler}) = P(\text{Sprinkler} \mid \text{Rain}) \cdot P(\text{Rain}) = 0.1 \cdot 0.2 = 0.02$
