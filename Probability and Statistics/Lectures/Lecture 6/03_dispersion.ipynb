# *Generating a lot of numbers*

```python
num_samples = 10000       # get rid of 'magic numbers'
```

```python
uniform_vals = np.random.uniform(low=0.0, high=1.0, size = num_samples)
```

```python
sns.displot(uniform_vals, bins = 20, kde = False)
plt.ylabel('Frequency')
plt.xlabel('Value')
plt.title("Uniform Values")
sns.despine(offset=10, trim=True)       # move axes away
plt.show()
```

```python
normal_vals = np.random.normal(loc=5.0, scale=10.0, size=num_samples)
# More values closer to 'loc'   .....Sort of like [1 5 5 5 5 5 7]
```

```python
sns.displot(normal_vals, bins = 20, kde = False)
plt.ylabel('Frequency')
plt.xlabel('Value')
plt.title("Mormal Values")
sns.despine(offset=10, trim=True)   # move axes away
```

```python
print("Uniform vals mean: ", np.mean(uniform_vals))
print("Normal vals mean: ", np.mean(normal_vals))
```

```python
print("Uniform vals variance: ", np.var(uniform_vals)) 
print("Normal vals variance: ", np.var(normal_vals))
```

```python
print("Uniform vals Standard Deviation: ", np.std(uniform_vals))
print("Normal vals Standard Deviation: ", np.std(normal_vals))
```

```python
normal_vals = np.random.normal(loc=5.0, scale=10.0, size= num_samples)

nv_mean = np.mean(normal_vals)
nv_std = np.std(normal_vals)
```

```python
sns.histplot(normal_vals, bins=20, kde=False)
plt.xlim(-60, 60)           # plot without it to see what happen to graph
plt.ylabel('Frequency')
plt.xlabel('Value')
plt.title("Normal Values")
sns.despine(offset=10, trim=True)     # moves axes away

# plot the SD line
x_c, y_c = ([nv_mean, nv_mean], [0, 1500])
plt.plot(x_c, y_c, color= 'green', linewidth = 2)

x_c, y_c = ([nv_mean + nv_std, nv_mean + nv_std], [0, 1500])
plt.plot(x_c, y_c, color= 'red', linewidth = 2)

x_c, y_c = ([nv_mean - nv_std, nv_mean - nv_std], [0, 1500])
plt.plot(x_c, y_c, color= 'red', linewidth = 2)

plt.show()
```

```python
normal_vals = np.random.normal(loc=5.0, scale=15.0, size= num_samples)      # changeing the scale to see  difference

nv_mean = np.mean(normal_vals)
nv_std = np.std(normal_vals)
```

```python
sns.histplot(normal_vals, bins=20, kde=False)
plt.xlim(-60, 60)

plt.ylabel('Frequency')
plt.xlabel('Value')
plt.title("Normal Values")
sns.despine(offset=10, trim=True)     # moves axes away

# plot the SD line
x_c, y_c = ([nv_mean, nv_mean], [0, 1500])
plt.plot(x_c, y_c, color= 'green', linewidth = 2)

x_c, y_c = ([nv_mean + nv_std, nv_mean + nv_std], [0, 1500])
plt.plot(x_c, y_c, color= 'red', linewidth = 2)

x_c, y_c = ([nv_mean - nv_std, nv_mean - nv_std], [0, 1500])
plt.plot(x_c, y_c, color= 'red', linewidth = 2)

plt.show()
```

```python
# Let's put both together
normal_vals = np.random.normal(loc=5.0, scale=10.0, size= num_samples)      # changeing the scale to see  difference
normal_vals_2 = np.random.normal(loc=5.0, scale=15.0, size= num_samples)      # changeing the scale to see  difference

def plot_hist(vals, label):
    nv_mean = np.mean(vals)
    nv_std = np.std(vals)

    sns.histplot(normal_vals, bins=20, kde=False)

    plt.xlim(-60, 60)

    plt.ylabel('Frequency')
    plt.xlabel('Value')
    plt.title(label)
    sns.despine(offset=10, trim=True)     # moves axes away

    # plot the SD line
    x_c, y_c = ([nv_mean, nv_mean], [0, 1500])
    plt.plot(x_c, y_c, color= 'green', linewidth = 2)

    x_c, y_c = ([nv_mean + nv_std, nv_mean + nv_std], [0, 1500])
    plt.plot(x_c, y_c, color= 'red', linewidth = 2)

    x_c, y_c = ([nv_mean - nv_std, nv_mean - nv_std], [0, 1500])
    plt.plot(x_c, y_c, color= 'red', linewidth = 2)

    plt.show()

plot_hist(normal_vals, "Less Variant")
plot_hist(normal_vals_2, "More Variant")
```

```python
med_incs = [x[0] for x in houses.data]

mean = np.mean(med_incs)
std = np.std(med_incs)

sns.histplot(med_incs, bins=20, kde=False)

# plt.xlim(-60, 60)

plt.ylabel('Frequency')
plt.xlabel('Median Income')
plt.title("Histogram of Median Incomes")
sns.despine(offset=10, trim=True)     # moves axes away

# plot the SD line
x_c, y_c = ([mean, mean], [0, 4000])
plt.plot(x_c, y_c, color= 'green', linewidth = 2)

x_c, y_c = ([mean + std, mean + std], [0, 4000])
plt.plot(x_c, y_c, color= 'red', linewidth = 2)

x_c, y_c = ([mean - std, mean - std], [0, 4000])
plt.plot(x_c, y_c, color= 'red', linewidth = 2)

plt.show()
```
