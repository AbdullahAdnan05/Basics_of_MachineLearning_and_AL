# **Data Visualization**

```python
# taking a list of data
y = [6, 7, 2, 4, 5]
```

```python
import matplotlib
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.plot(y)
plt.show()
```

```python
# create a figure 
plt.figure(figsize = (10, 5))

# add a line renderer with legend and line thickness
# plt.plot(y)
plt.plot(y, label='value', linewidth=4)
plt.legend()

# also need to make the figure readable -- very important from stats perspective
plt.title("First Example Plot")
plt.xlabel("x vals (seconds)")
plt.ylabel("y vals (meters)")
plt.grid()

plt.show()
```

```python
# # Plotting Weights
# Create a figure 
plt.figure(figsize=(10, 5))

# add a line renderer with legend and line thickness
x = range(da['BMXWT'].size)
y = da['BMXWT']

plt.scatter(x, y, marker='x', s = 0.1)      # marker 'x' tell to make size small

# also need to make the figure readable -- very important from stats perspective!
plt.title("Weights of Individuals")
plt.xlabel("ID of Individual")
plt.ylabel("Weight (kg)")

plt.show()
```

```python
sum(y)       # it will give nan(not a number) due to sum(y) which mean there will some missing value in the column
```

```python
# it will not remove values from data frame instead it will remove from the variable you create for the column
y = y.dropna()      # just remove the values with NaN 
```

```python
sum(y)
```

```python
avg = sum(y) / len(y)
print("total averge weight = ", avg)
```
