# **Data Visualization**

```python
# taking a list of data
y = [6, 7, 2, 4, 5]
```

```python
import matplotlib
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.plot(y)
plt.show()
```

```python
# create a figure 
plt.figure(figsize = (10, 5))

# add a line renderer with legend and line thickness
plt.plot(y)

# also need to make the figure readable -- very important from stats perspective
plt.title("First Example Plot")
plt.xlabel("x vals (seconds)")
plt.ylabel("y vals (meters)")
plt.grid()

plt.show()
```

```python
## Ploting Weights
# create a figure 
plt.figure(figsize=(10, 5))

# add a line renderer with legend and line thickness
x = range(da["BMXWT"].size)
y = da['BMXWT']

plt.scatter(x,y,marker= 'x', s = 0.1)

# also need to make the figure readable -- very important from stats perpective! 
plt.title("Weights of individuals")
plt.xlabel("ID of Individual")
plt.ylabel("Weight (kg)")

plt.show()
```
