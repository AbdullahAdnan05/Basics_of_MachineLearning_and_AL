# **More Generalize Way to right Newton's Method:**

```python
def new_val(x, f, fp):
    return x - (f(x)/ fp(x))
```

```python
def root_81(x):
    return x **2 -81
```

```python
def root_81_p(x):
    """Derviative of the root_81 function"""
    return 2*x
```

```python
initial_value = 7
```

```python
x = new_val(initial_value, root_81, root_81_p)
print(x)
```

```python
x = new_val(x, root_81, root_81_p)
print(x)
```

```python
x = new_val(x, root_81, root_81_p)
print(x)
```
