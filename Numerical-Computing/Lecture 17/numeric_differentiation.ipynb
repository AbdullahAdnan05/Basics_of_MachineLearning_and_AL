# **Numeric Differentiation**

Given the following data:

x | y

1 | 24

2 | 120

3 | 536

7 | 720

Calculate the slope at x=8

```python
h = 2
x = 8
x0 = 1
y0 = 24

dy0= 96
d2y0 = 120
d3y0 = 48

p = (x-x0)/h
print(h ,x, p)
```

```python
def fact(n):
    val = 1
    for i in range(1,n+1):
        val *= i
    return val
```

```python
def inter_poly(x, x0, h, dy0, d2y0, d3y0):
    # Compute p first
    p = (x-x0)/h
    
    return                      y0 + \
          p                  * dy0 + \
          p*(p-1)/fact(2)    * d2y0 + \
          p*(p-1)*(p-2)/fact(3)  * d3y0
```

```python
x = 8
inter_poly(x, x0, h, dy0, d2y0, d3y0)
```

```python
x_m = 7.5
y_m = inter_poly(x_m, x0, h, dy0, d2y0, d3y0)
print(y_m)
```

```python
x_p = 8.5
y_p = inter_poly(x_p, x0, h, dy0, d2y0, d3y0)
print(y_p)
```

```python
(y_p - y_m) / (x_p - x_m)
```

# **Differential**

Let's do it directly!

```python
def numeric_diff(x ,x0, h, dy0, d2y0, d3y0):
    # Compute p first
    p = (x-x0)/h
    print("p = ",p)
    
    return (
             dy0 + \
             ((2*p)-1)  / fact(2)           * d2y0 + \
             (3*p**2 - 6*p +2 ) / fact(3)  * d3y0   \
    ) / h
```

```python
numeric_diff(8, x0, h, dy0, d2y0, d3y0)
```
