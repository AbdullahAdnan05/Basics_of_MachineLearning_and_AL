# **1. Number and Types**


| **Set Name**       | **Symbol**        | **Description**                                                                 |
|---------------------|-------------------|---------------------------------------------------------------------------------|
| **Naturals**        | \( N \)           | \( N = \{1, 2, 3, 4, \dots\} \)                                                |
| **Wholes**          | \( W \)           | \( W = N \cup \{0\} \)                                                         |
| **Integers**        | \( Z \)           | \( Z = W \cup \{-1, -2, -3, \dots\} \)                                         |
| **Rationals**       | \( Q \)           | \( Q = \left\{\frac{p}{q} : p \in Z, q \in Z \setminus \{0\}\right\} \)        |
| **Irrationals**     | \( I \)           | \( I \) is the set of real numbers not expressible as a fraction of integers.  |
| **Reals**           | \( R \)           | \( R = Q \cup I \)                                                             |
| **Complex Numbers** | \( C \)           | \( C = \{a + bi : a, b \in R, i = \sqrt{-1}\} \)                               |


```python
import numpy as np
vector_row = np.array([[1, -5, 3, 2, 4]])
vector_column = np.array([[1],
                          [2],
                          [3],
                          [4]])
print(vector_row.shape)
print(vector_column.shape)
```

# **2. Floating Point Numbers**

```python
import sys
sys.float_info
```

```python
import numpy as np

# Spacing tells us how far "next largest/smallest" number is
np.spacing(100_000_000)
```

```python
np.spacing(100.2)       # notice smaller spacing at this point
```

```python
print(  np.spacing(1e4)  )
print(  np.spacing(1e-4)  )
```

```python
num = 1e9
# num = 100.0
spacing = np.spacing(num)
print(num)
print(spacing)

num_next = num + spacing
print(num_next)
print("This is same:    ", num == num_next)


num_closer = num + (spacing/2)
print(num_closer)
print("Closer is same:  ", num == num_closer)
```
