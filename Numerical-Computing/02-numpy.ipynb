# **2.Misc Operations with Numpy**

```python
import math
def basic_sigmoid(x):
    """
    Compute sigmoid of x.

    Arguments:
    x -- A    scalar
    
    Return:
    s -- sigmoid(x)
    """
    
    s = 1./(1. + math.e ** (-x))
    
    return s
```

```python
help(basic_sigmoid)
```

```python
basic_sigmoid(-1)
```

```python
basic_sigmoid(0)
```

```python
basic_sigmoid(6)
```

```python
x = [-1, 0, 3]
basic_sigmoid(x)        # why
```

```python
# how to solve the above cell
import numpy as np

x = [-1, 0, 3]
x = np.array(x)
basic_sigmoid(x)
```

# **2.2. Normalizing Rows and Columns**

Often, we wish to normalize value of a row to fall within a specific range.

```python
x = np.array([   [0, 3 , 4],
                 [1, 6 , 4] ])
```

```python
x_norm = np.amax(x, axis=1)         # get row-wise max
print(x_norm)
```

```python
x / x_norm      #  this does not work. Why?
```

```python
x_norm = x_norm.reshape(2,1)
```

```python
x_norm
```

```python
x/x_norm
```

```python
# We can also use another normalization method
x = np.array([ [ 0, 3, 4 ],
              [ 1, 6, 4 ],])

# no need to reshape again
x_norm = np.linalg.norm(x, ord = 2, axis = 1, keepdims = True)
x / x_norm
```

# **2.4. Vectorization**

```python
dim = 100            # try increasing this after running experiment
```

```python
A = np.random.rand(dim, dim)
B = np.random.rand(dim, dim)
```

```python
A[0].size, A[1].size
```

```python
# Matrix Addition
def add_arrays(A, B):
    C = np.zeros((A[0].size, A[1].size))
    
    for i in range(A[0].size):
        for j in range(A[1].size):
            C[i, j] = A[i, j]  +  B[i, j]
    
    return C
```

```python
%time C = add_arrays(A, B)      # look at the differenc which you made
```

```python
# And what python made
%time C = A + B
```

```python
import time

# Non-vectorized time
start = time.time()

C = add_arrays(A, B)

end = time.time()
non_vec_time = end - start 


# Vectorized time
start = time.time()

C = A + B

end = time.time()
vec_time = end - start
```

```python

vec_time / non_vec_time * 100
```
