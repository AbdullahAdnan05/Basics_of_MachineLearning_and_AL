# **1.3. Vector Norm**

The **norm** of a vector is a measure of its length. There are many ways of defining the length of a vector depending on the metric used (i.e., the distance formula chosen). So let's take a look at same distance formulae first.

# **1.3.1. Euclidean Distance**

Euclidean distance is what you are already familiar with. The Euclidean distance between two points (x1, y1) and (x2, y2) is given as

 :
                       $d = √[(x2 – x1)^2 + (y2 – y1)^2]$

Visually, this would look like (see notebook for this).

If we take any vector and take it's x- and y-corridinates as its components, we can redefine the above like so:

This length is then called the L-2 Norm of the vector -- stuff you know as the magnitude. The specific formula is :

:
                         $  |x|2=√(∑vi^2)  $ where ∑i=1

where vi are the elment of the vector v.

# **1.5.2. Determinants**

```python
from numpy.linalg import det

M = np.array(
            [  [0, 2, 1, 3],
               [3, 2, 8, 1],
               [1, 0, 0, 3],
               [0, 3, 2, 1],  ]
            )

print('M:\n', M)

print('Determinat: %1.f' % det(M))

I = np.eye(4)   # eye-identity matrix

print('I:\n', I)
print('M=I:\n', np.dot(M, I))
```

# **1.5.3. Inverse, Singular, Ill-conditioned**

When you multiply something by a matrix and you want to get back, you multiply the result with the inverse of the original matrix. This is similar to multiplying by the reciprocal to reverse the effect of a multiplication.

```python
from numpy.linalg import inv

print('Inv M:\n', inv(M))
```

```python
P = np.array([[0, 1, 0],
              [0, 0, 0],
              [1, 0, 1],])     # Phas no inverse since it is singular.

print('det(P):', det(P))

inv(P)
```

So, multiplying by a singular matrix is sort of like multiplying by 0. It's not reversible.

Ill-conditioned matrices are not singular but their determinant is very close to 0. This makes them problematic just like multiplication by very small numbers can lead to overflow, underflow etc.

# **1.5.4. Rank, Full Rank, Condition Number, Augmented Matrix**

Rank, rank(A), is the number of linerly independent columns or rows of A.

Full rank if rank(A) = min(m,n) for an m X n matrix.

Condition Number: How close is a matrix to being ill-conditoned. Higher is closer to being singular.

Augmented matrix:[A|y] where A is a matrix and y is a vector. y is called new if it cannot be expressed as a combination of columns of A. In this case, rank([A|y]) = rank(A) + 1 (because we just added a new independent column!)

```python
from numpy.linalg import cond, matrix_rank

A = np.array([[1,1,0],
              [0,1,0],
              [1,0,1],])

print('Condition number:\n', cond(A))

print('Rank:\n', matrix_rank(A))

y = np.array([[1, 2, 1]]).T     # need a column vector. Easier to create it using transpose

A_y = np.concatenate((A, y), axis= 1)       # Notice the axis
print('Augmented matrix:\n', A_y)
```

# **Gauss_Seidal Method**

$8x_1 + 3x_2 - 3x_3  = 14$

$-2x_1 - 8x_2 - 5x_3  = 5$

$3x_1 + 5x_2 + 10x_3  = -8$

*Gauss_Seidal Method:* It is a bit different in Math term here it mean simple so do how does 'Gauss_Seidal Method' work for better understanding

```python
# to solve with this specific method it is important to have value in diagonal greater then the other two values knows as "diagonally dominant"
a = [
    [ 8, 3, -3],
    [-2,-8,  5],
    [ 3, 5,  10],
]

# Find diagonal coefficients
diag = np.diag(np.abs(a))
print(diag)
```

```python
np.sum(np.abs(a), axis=1)           # add all values of rows in matrix
```

```python
# Find row sum without diagonal

off_diag = np.sum(np.abs(a), axis=1) - diag         # subtract the diagnal from each row
print(off_diag)
```

```python
if np.all(diag > off_diag):
    print('matrix is diagonally dominant')      # will converge (sufficient condition for convergence in iterative methods )
else:
    print('Not diagonially dominant')           # may or may not converge (because diagonal dominance is a sufficient but not necessary condition)
```

Re-arrange the original:

$8x_1 + 3x_2 - 3x_3  = 14$

$-2x_1 - 8x_2 - 5x_3  = 5$

$3x_1 + 5x_2 + 10x_3  = -8$

as:

*Gauss_Seidal Method:* It is a bit different in Math term here it mean simple so do how does 'Gauss_Seidal Method' work for better understanding

$x_1 = (14 - 3x_2 + 3x_3)/8$

$x_2 = (5 +2x_1 - 5x_3)/(-8) $

$x_3  = (-8 - 3x_1 - 5x_2)/10$

```python
# Let's apply the iterative method starting at 0

x1 = 0
x2 = 0
x3 = 0
epsilon = 0.001     # also try better precision
converge = False

x_old = np.array([x1, x2, x3])
```

```python
max_iteration = 50
```

```python
print('Iterative results')
print('K    x1        x2         x3')

for k in range(1, max_iteration):
    
    x1 = (14 - 3*x2 + 3*x3)  / 8
    x2 = (5 + 2*x1 - 5*x3)  / (-8)
    x3 = (-8 - 3*x1 - 5*x2)  / 10
    
    x = np.array([x1, x2, x3])
    
    print("%2d   %.4f   %.4f   %.4f " %  (k ,x1,  x2,  x3))
    
    # check if it is smaller than threhold
    dx = np.sqrt(np.dot(x-x_old, x-x_old))
    
    if dx < epsilon:
        converge = True
        print('converged!')
        break
    
    # assign the latest x value to the old value
    x_old = x
    
if not converge:
    print('Not converged, increase the # of iterations')
```
