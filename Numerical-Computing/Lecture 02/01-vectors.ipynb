# **1. Vector/Matrix Definition and Usage**

**1.1. Vector Representation**

```python
class Vector:
    def __init__(self, x = 0.0, y = 0.0):
        self.x = x
        self.y = y
        
    def __str__(self):
        return "[{}, {}]".format(str(self.x), str(self.y))
```

```python
a = Vector(2, 4)
```

```python
print(a)
```

```python
b = Vector(5, 2)
print(b)
```

```python
def add(self, b):
    c = Vector()
    c.x = self.x + b.x
    c.y = self.y + b.y
    
    return c
Vector.add = add
```

```python
c = a.add(b)
print(c)
```

```python
def mul(self, s):
    
    return Vector(s * self.x, s * self.y)
Vector.mul = mul
```

```python
d = a.mul(2)
print(d)
```

```python
def sub(self, b):
    # We use the definition of vector subraction
    #   instead of defining something new
    
    return self.add( b.mul(-1) )

Vector.sub = sub
```

```python
d_min_b = d.sub(b)
print(d_min_b)
```

```python
def dot_product(self, b):
    c = Vector()
    c.x = self.x * b.x
    c.y = self.y * b.y
    dot = c.x+c.y
    
    return dot 

Vector.dot_product = dot_product
```

```python
c = a.dot_product(b)
print("Dot product of a and b = ",c)
```

```python
def cross_product(self, b):
    c = (self.x)*(b.y) - (self.y) * (b.x)
    return c
Vector.cross_product = cross_product
```

```python
c = a.cross_product(b)
print("Cross product of a and b = ", c)
```
