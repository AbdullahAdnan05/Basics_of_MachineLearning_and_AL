# **1. Vector/Matrix Definition and Usage**

**1.1. Vector Representation**

```python
class Vector:
    def __init__(self, x = 0.0, y = 0.0):
        self.x = x
        self.y = y
        
    def __str__(self):
        return "[{}, {}]".format(str(self.x), str(self.y))
```

```python
a = Vector(2, 4)
```

```python
print(a)
```

```python
b = Vector(5, 2)
print(b)
```

```python
def add(self, b):
    c = Vector()
    c.x = self.x + b.x
    c.y = self.y + b.y
    
    return c
Vector.add = add
```

```python
c = a.add(b)
print(c)
```

```python
def mul(self, s):
    
    return Vector(s * self.x, s * self.y)
Vector.mul = mul
```

```python
d = a.mul(2)
print(d)
```

```python
def sub(self, b):
    # We use the definition of vector subraction
    #   instead of defining something new
    
    return self.add( b.mul(-1) )

Vector.sub = sub
```

```python
d_min_b = d.sub(b)
print(d_min_b)
```

# **Matrix Representation (Dense)**

```python
class Matrix:
    def __init__(self, dims, fill):         # fill is use for value of matrix
        self.rows = dims[0]                 # dims is a tuple
        self.cols = dims[1]
        
        self.A = [
            [fill] * self.cols              # for each row, this many columns
            for i in range(self.rows)       # create this many rows
        ]
```

```python
m = Matrix((3,4),2.0)
```

```python
print(m)
```

```python
def __str__(self):
    rows = len(self.A)      # Get the first dimension
    ret = ''
    
    for i in range(rows):
        cols = len(self.A[i])
        
        for j in range(cols):
            ret += str(self.A[i][j]) + "\t"
        ret += "\n"
        
    return ret


Matrix.__str__ = __str__
```

```python
# recall that we can get values from our matrix using indices
def get(self, i, j):
    
    # Error checking
    if i < 0 or i > self.rows:
        raise ValueError("Row index out of range.")
    if j < 0 or j > self.cols:
        raise ValueError("Column index out of range.")
    
    # Value return
    return self.A[i][j]

Matrix.get = get
```

# **1.2.2.Matrix Representation(Sparse)**

```python
class Matrix:
    def __init__(self, dims):
        self.rows = dims[0]
        self.cols = dims[1]
        self.vals = {}
        
        # Let's assume for a minute that fill is 0
        
    # obviously need a need __str__ here .....
```

```python
def __str__(self):
    rows = self.rows    # Get the first dimension
    ret = ''
    
    for i in range(rows):
        cols = self.cols
        
        for j in range(cols):
            
            if (i, j) in self.vals:
                ret += str(self.vals[(i,j)]) + "\t"
            else:
                ret += str(0.0) + "\t"
                
        
        ret += "\n"
        
    return ret


Matrix.__str__ = __str__
```

```python
def set(self, i,j, val):
    self.vals[(i,j)] = val
    
Matrix.set = set
```

```python
# spare implementation of get
def get(self, i, j):
    
    # Error checking
    if i<0 or i > self.rows:
        raise ValueError("Row index out of range.")
    if j<0 or j > self.cols:
        raise ValueError("Columns index out of range.")
    
    # value return
    if (i, j) in self.vals:
        return self.vals[(i, j)]
        
    return 0.0
    

Matrix.get = get
```
